```python
import pandas as pd
import statsmodels.api as sm
import statsmodels.formula.api as smf
from statsmodels.stats.outliers_influence import variance_inflation_factor as vif
#from statsmodels.graphics.regresssionplots import *
#from sklearn import datasets, lienar_model
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Boston.csv", header=0, index_col=0)
```

```python
df.head()
```

```python
df.columns
```

```python
#criação do modelo de regressão simples de medv somente pela feature lstat
lm = smf.ols('medv~lstat', data=df).fit()
```

```python
print(lm.summary())
```

```python
print(lm.conf_int())
print("==============")
print(lm.params)
```

```python
Plotando a linha de regressão linear e seus resíduos.
```

```python
x_new = pd.DataFrame({'lstat': [df.lstat.min(), df.lstat.max()]})
preds = lm.predict(x_new)
plt.figure(figsize=(15,5))
plt.scatter(df["lstat"], df["medv"], c="black")
plt.plot(x_new, preds, c='red')
plt.show()
```

```python
plt.figure(figsize=(15,8))
plt.subplot(2,1,1)
plt.plot(df.lstat, lm.predict(), 'ro')
plt.subplot(2,1,2)
plt.plot(lm.predict(), lm.resid, 'go')
```

```python
lm2 = smf.ols("medv~lstat+ I(lstat ** 2)", data=df).fit()
print(lm2.summary())
print("==============================================================================")
```

```python
plt.figure(figsize=(15,8))
plt.subplot(2,1,1)
plt.plot(df.lstat, lm2.predict(), 'ro')
plt.subplot(2,1,2)
plt.plot(lm2.predict(), lm2.resid, 'go')
```

```python
table = sm.stats.anova_lm(lm, lm2)
print(table)
```
